## 3. Operators

### 3.1 Arithmetic Operators


```python
a, b = 17, 5

print("Addition:       ", a + b)   # 22
print("Subtraction:    ", a - b)   # 12
print("Multiplication: ", a * b)   # 85
print("Division:       ", a / b)   # 3.4  (always float)
print("Floor Division: ", a // b)  # 3    (truncates decimal)
print("Modulus:        ", a % b)   # 2    (remainder)
print("Exponentiation: ", a ** b)  # 1419857
```

### 3.2 Bitwise Operators

```python
x, y = 0b1100, 0b1010   # 12 and 10 in binary

print(f"x         = {x:04b}  ({x})")
print(f"y         = {y:04b}  ({y})")
print(f"x & y     = {(x & y):04b}  ({x & y})   AND")
print(f"x | y     = {(x | y):04b}  ({x | y})   OR")
print(f"x ^ y     = {(x ^ y):04b}  ({x ^ y})   XOR")
print(f"~x        = {(~x)}            NOT")
print(f"x << 1    = {(x << 1):05b}  ({x << 1})  Left shift")
print(f"x >> 1    = {(x >> 1):04b}  ({x >> 1})   Right shift")
```

### 3.3 Assignment Operators

```python
n = 10
print("Start:", n)
n += 5;  print("+=5  →", n)
n -= 3;  print("-=3  →", n)
n *= 2;  print("*=2  →", n)
n //= 3; print("//=3 →", n)
n **= 2; print("**=2 →", n)
n %= 7;  print("%=7  →", n)
```

### 3.4 Comparison Operators

```python
a, b = 10, 20
print(a == b)   # Equal
print(a != b)   # Not equal
print(a <  b)   # Less than
print(a >  b)   # Greater than
print(a <= b)   # Less than or equal
print(a >= b)   # Greater than or equal

# Chained comparisons (Pythonic!)
x = 15
print(10 < x < 20)    # True
print(1 < 2 < 3 < 4)  # True
```

### 3.5 Boolean Operators

```python
t, f = True, False

print(t and f)   # False — both must be True
print(t or  f)   # True  — at least one True
print(not  t)    # False — negation

# Short-circuit evaluation
x = None
result = x or "default"
print(result)   # "default"

y = "hello"
result = y and y.upper()
print(result)   # "HELLO"
```

### 3.6 Identity and Membership Operators

```python
# Identity: is / is not  (same object in memory)
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)       # True  — same object
print(a is c)       # False — equal values, different objects
print(a is not c)   # True

# Membership: in / not in
fruits = ["apple", "banana", "cherry"]
print("apple"  in fruits)       # True
print("mango"  not in fruits)   # True

sentence = "Python is awesome"
print("Python" in sentence)     # True
print("Java"   in sentence)     # False
```

## 4. Data Structures & Types

### 4.1 Simple (Scalar) Types


```python
# int
age = 25
big = 1_000_000        # underscores for readability
hexa = 0xFF            # hex literal
print(age, big, hexa)

# float
pi   = 3.14159
sci  = 2.5e-3          # scientific notation
print(pi, sci)

# complex
z = 3 + 4j
print(z, z.real, z.imag, abs(z))

# bool (subclass of int)
print(True + True)     # 2
print(False * 100)     # 0

# str
greeting = "Hello, World!"
multiline = """Line 1
Line 2
Line 3"""
print(greeting.upper(), greeting.lower())
print(greeting.replace("World", "Python"))
print(greeting.split(", "))
print(len(greeting))
```

### 4.2 Compound Types — Lists

```python
# Lists — ordered, mutable, allows duplicates
fruits = ["apple", "banana", "cherry", "apple"]
print(fruits)

fruits.append("mango")          # add to end
fruits.insert(1, "blueberry")   # insert at index
fruits.remove("apple")          # remove first occurrence
popped = fruits.pop()           # remove & return last item
fruits.sort()                   # in-place sort

print(fruits)
print("Popped:", popped)
print("Length:", len(fruits))
print("Count 'apple':", fruits.count("apple"))
print("Index of 'banana':", fruits.index("banana"))
```

### 4.3 Compound Types — Tuples

```python
# Tuples — ordered, IMMUTABLE, allows duplicates
coords = (10.5, 20.3)
rgb    = (255, 128, 0)
single = (42,)           # note trailing comma for single-element tuple

print(coords, rgb, single)
print("x:", coords[0], "y:", coords[1])

# Tuple unpacking
x, y = coords
print(f"Unpacked: x={x}, y={y}")

r, g, b = rgb
print(f"RGB: r={r}, g={g}, b={b}")

# Swapping with tuples
a, b = 1, 2
a, b = b, a
print(f"Swapped: a={a}, b={b}")

# Named tuple
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 7)
print(p, p.x, p.y)
```

### 4.4 Compound Types — Dictionaries

```python
# Dicts — key-value pairs, ordered (3.7+), mutable
person = {
    "name": "Alice",
    "age":  30,
    "city": "Toronto"
}

# Access
print(person["name"])
print(person.get("email", "N/A"))  # safe get with default

# Modify
person["age"] = 31
person["email"] = "alice@example.com"
del person["city"]

# Iterate
for key, value in person.items():
    print(f"  {key}: {value}")

print("Keys:  ", list(person.keys()))
print("Values:", list(person.values()))

# Merging (Python 3.9+)
defaults = {"theme": "dark", "lang": "en"}
settings = {**defaults, "lang": "fr", "font": "Mono"}
print(settings)
```

### 4.5 Collections — Sets

```python
# Sets — unordered, unique elements, mutable
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7}

print("Union:        ", a | b)
print("Intersection: ", a & b)
print("Difference:   ", a - b)
print("Sym Diff:     ", a ^ b)

# Useful for deduplication
dupes = [1, 2, 2, 3, 3, 3, 4]
unique = list(set(dupes))
print("Unique:", unique)

# frozenset — immutable set
fs = frozenset([1, 2, 3])
print(fs)
```

### 4.6 Collections Module

```python
from collections import Counter, defaultdict, deque, OrderedDict

# Counter — count occurrences
words = ["cat", "dog", "cat", "bird", "dog", "cat"]
c = Counter(words)
print("Counter:", c)
print("Most common:", c.most_common(2))

# defaultdict — dict with default value factory
dd = defaultdict(list)
dd["fruits"].append("apple")
dd["fruits"].append("banana")
dd["vegs"].append("carrot")
print("defaultdict:", dict(dd))

# deque — double-ended queue (efficient prepend/append)
dq = deque([1, 2, 3])
dq.appendleft(0)
dq.append(4)
dq.rotate(1)   # rotate right by 1
print("deque:", dq)
```

## 5. Indexing & Slicing

Python uses **zero-based indexing**. Negative indices count from the end.


```python
s = "Python Programming"
lst = [10, 20, 30, 40, 50, 60, 70]

# ── Indexing ──────────────────────────────────────────
print("s[0]  =", s[0])     # 'P'
print("s[-1] =", s[-1])    # 'g'
print("lst[2] =", lst[2])  # 30
print("lst[-2]=", lst[-2]) # 60

# ── Slicing: [start : stop : step] ───────────────────
print(s[0:6])     # 'Python'
print(s[7:])      # 'Programming'
print(s[:6])      # 'Python'
print(s[::2])     # every 2nd character
print(s[::-1])    # reverse string

print(lst[1:5])   # [20, 30, 40, 50]
print(lst[::2])   # [10, 30, 50, 70]
print(lst[::-1])  # reversed list

# ── Slice assignment ──────────────────────────────────
lst[1:3] = [200, 300]
print("After slice assignment:", lst)
```

## 6. Conditional Statements


```python
# ── if / elif / else ─────────────────────────────────
score = 78

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} → Grade: {grade}")

# ── Ternary (conditional expression) ─────────────────
age = 20
status = "adult" if age >= 18 else "minor"
print(status)

# ── match / case (Python 3.10+) ───────────────────────
command = "quit"
match command:
    case "start":
        print("Starting...")
    case "stop" | "quit":
        print("Stopping/Quitting")
    case _:
        print("Unknown command")
```

## 7. Loops


```python
# ── for loop ─────────────────────────────────────────
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# ── range() ───────────────────────────────────────────
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 10, 2):   # start, stop, step
    print(i, end=" ")
print()

# ── enumerate() ───────────────────────────────────────
for idx, fruit in enumerate(fruits, start=1):
    print(f"{idx}. {fruit}")

# ── zip() ─────────────────────────────────────────────
names  = ["Alice", "Bob", "Carol"]
scores = [95, 82, 91]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

```python
# ── while loop ───────────────────────────────────────
count = 0
while count < 5:
    print(count, end=" ")
    count += 1
print()

# ── break, continue, else ────────────────────────────
for i in range(10):
    if i == 3:
        continue    # skip 3
    if i == 7:
        break       # stop at 7
    print(i, end=" ")
print()

# for-else: else runs if loop completes without break
for n in range(2, 10):
    for factor in range(2, n):
        if n % factor == 0:
            break
    else:
        print(f"{n} is prime", end="  ")
print()
```

### 8.3 Default Arguments

```python
def power(base, exponent=2):
    """Raise base to exponent (default: square)."""
    return base ** exponent

print(power(3))       # 9  — uses default
print(power(3, 3))    # 27
print(power(2, 10))   # 1024

def connect(host, port=5432, ssl=True):
    print(f"Connecting to {host}:{port} (SSL={ssl})")

connect("db.example.com")
connect("db.example.com", port=3306, ssl=False)
```

### 8.4 `*args` and `**kwargs`

```python
# *args — variable positional arguments (tuple)
def add_all(*args):
    print(f"args = {args}")
    return sum(args)

print(add_all(1, 2, 3))
print(add_all(10, 20, 30, 40, 50))

# **kwargs — variable keyword arguments (dict)
def display_info(**kwargs):
    for key, val in kwargs.items():
        print(f"  {key}: {val}")

display_info(name="Alice", age=30, city="Toronto")

# Combining all parameter types
def full_example(required, *args, default="hi", **kwargs):
    print(f"required={required}, args={args}, default={default}, kwargs={kwargs}")

full_example("must", 1, 2, 3, default="hey", x=10, y=20)

# Unpacking when calling
nums = [1, 2, 3]
info = {"name": "Bob", "age": 25}
print(add_all(*nums))
display_info(**info)
```

## 9. Comprehensions & Generators

### 9.1 List Comprehensions


```python
# [expression  for item in iterable  if condition]

squares       = [x**2 for x in range(10)]
even_squares  = [x**2 for x in range(10) if x % 2 == 0]
words         = ["hello", "world", "python"]
upper_words   = [w.upper() for w in words]

print(squares)
print(even_squares)
print(upper_words)

# Nested list comprehension
matrix    = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [n for row in matrix for n in row]
print(flattened)

# 2D grid
grid = [(x, y) for x in range(3) for y in range(3)]
print(grid)
```

### 9.2 Set Comprehensions

```python
# {expression for item in iterable if condition}
sentence = "the quick brown fox jumps over the lazy dog"
unique_chars  = {ch for ch in sentence if ch != " "}
word_lengths  = {len(w) for w in sentence.split()}

print(sorted(unique_chars))
print(word_lengths)   # set of unique lengths — no duplicates
```

### 9.3 Dict Comprehensions

```python
# {key_expr: val_expr for item in iterable if condition}
squares_dict  = {x: x**2 for x in range(1, 8)}
print(squares_dict)

# Invert a dictionary
original  = {"a": 1, "b": 2, "c": 3}
inverted  = {v: k for k, v in original.items()}
print(inverted)

# Filter dict items
prices    = {"apple": 1.5, "banana": 0.5, "cherry": 3.0, "mango": 2.5}
expensive = {k: v for k, v in prices.items() if v > 1.0}
print(expensive)
```

### 9.4 Generators & Generator Expressions

```python
# Generator expression — lazy evaluation, memory-efficient
# (expression for item in iterable if condition)

gen = (x**2 for x in range(10))
print(type(gen))           # <class 'generator'>
print(next(gen))           # 0
print(next(gen))           # 1
print(list(gen))           # consume the rest

# Useful for large datasets
total = sum(x**2 for x in range(1_000_000))   # no list created!
print(f"Sum of squares 0–999999: {total:,}")
```

### 9.5 Generator Functions with `yield`

```python
# yield pauses function, saves state, resumes on next()
def countdown(n):
    print("Starting countdown!")
    while n > 0:
        yield n
        n -= 1
    print("Blastoff!")

for num in countdown(5):
    print(num, end=" ")
print()

# Fibonacci generator
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fib = fibonacci()
print([next(fib) for _ in range(12)])
```

### 9.6 Infinite Number Generator

```python
import itertools

# count — infinite counter
counter = itertools.count(start=1, step=2)
odds_10 = [next(counter) for _ in range(10)]
print("First 10 odd numbers:", odds_10)

# cycle — infinite cycle over iterable
cycler = itertools.cycle(["red", "green", "blue"])
colors = [next(cycler) for _ in range(7)]
print("Cycling colors:", colors)

# Custom infinite range generator
def infinite_range(start=0, step=1):
    n = start
    while True:
        yield n
        n += step

gen = infinite_range(0, 5)
multiples_of_5 = [next(gen) for _ in range(8)]
print("Multiples of 5:", multiples_of_5)

# islice to safely consume infinite generators
print(list(itertools.islice(infinite_range(0, 3), 10)))
```

## 10. Classes & Objects

### 10.1 Defining Classes


```python
class Animal:
    # Class variable (shared by all instances)
    kingdom = "Animalia"

    def __init__(self, name, sound):
        # Instance variables
        self.name  = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says '{self.sound}'"

    def __repr__(self): # representation
        return f"Animal(name={self.name!r}, sound={self.sound!r})"

    def __str__(self):
        return self.name


dog = Animal("Dog", "Woof")
cat = Animal("Cat", "Meow")

print(dog.speak())
print(cat.speak())
print(repr(dog))
print(str(cat))
print(f"Kingdom: {Animal.kingdom}")
```

### 10.2 Inheritance

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make  = make
        self.model = model
        self.year  = year

    def description(self):
        return f"{self.year} {self.make} {self.model}"

    def fuel_type(self):
        return "Gasoline"


class ElectricVehicle(Vehicle):
    def __init__(self, make, model, year, battery_kwh):
        super().__init__(make, model, year)    # call parent __init__
        self.battery_kwh = battery_kwh

    def fuel_type(self):                       # method override
        return "Electric"

    def range_estimate(self, km_per_kwh=6):
        return self.battery_kwh * km_per_kwh


class HybridVehicle(Vehicle):
    def fuel_type(self):
        return "Hybrid (Gas + Electric)"


# Usage
car     = Vehicle("Toyota", "Camry", 2022)
tesla   = ElectricVehicle("Tesla", "Model 3", 2024, 75)
prius   = HybridVehicle("Toyota", "Prius", 2023)

for v in [car, tesla, prius]:
    print(f"{v.description()} — Fuel: {v.fuel_type()}")

print(f"Tesla range: ~{tesla.range_estimate()} km")

# isinstance / issubclass
print(isinstance(tesla, Vehicle))         # True
print(isinstance(tesla, ElectricVehicle)) # True
print(issubclass(ElectricVehicle, Vehicle)) # True
```

### 10.3 Operator Overloading

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):          # v1 + v2
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):          # v1 - v2
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):         # v * 3
        return Vector(self.x * scalar, self.y * scalar)

    def __rmul__(self, scalar):        # 3 * v
        return self.__mul__(scalar)

    def __abs__(self):                 # abs(v) — magnitude
        return (self.x**2 + self.y**2) ** 0.5

    def __eq__(self, other):           # v1 == v2
        return self.x == other.x and self.y == other.y

    def __len__(self):                 # len(v) — dimension
        return 2


v1 = Vector(2, 3)
v2 = Vector(1, 4)

print(v1 + v2)         # Vector(3, 7)
print(v1 - v2)         # Vector(1, -1)
print(v1 * 3)          # Vector(6, 9)
print(3 * v2)          # Vector(3, 12)
print(abs(v1))         # 3.605...
print(v1 == Vector(2, 3))  # True
print(len(v1))         # 2
```

### 10.4 Properties, Class Methods, Static Methods

```python
class Circle:
    PI = 3.14159265358979

    def __init__(self, radius):
        self._radius = radius   # convention: _ = private

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value

    @property
    def area(self):
        return self.PI * self._radius ** 2

    @property
    def circumference(self):
        return 2 * self.PI * self._radius

    @classmethod
    def from_diameter(cls, diameter):
        """Alternative constructor."""
        return cls(diameter / 2)

    @staticmethod
    def is_valid_radius(value):
        return value >= 0

    def __repr__(self):
        return f"Circle(r={self._radius})"


c = Circle(5)
print(c.area, c.circumference)

c.radius = 10
print(c)

c2 = Circle.from_diameter(14)
print(c2)

print(Circle.is_valid_radius(-1))   # False
```

### 11.7 Building Your Own Module

```python
# Save as ./mymath.py (simulated here as a string)
module_code = '''
"""mymath — simple custom math utilities."""

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def primes_up_to(limit):
    return [n for n in range(2, limit+1) if is_prime(n)]

if __name__ == "__main__":
    # Only runs when executed directly, not when imported
    print("Testing mymath module")
    print(factorial(5))
    print(primes_up_to(30))
'''

with open("./mymath.py", "w") as f:
    f.write(module_code)

# Import and use
import sys; sys.path.insert(0, "/tmp")
import mymath

print(mymath.factorial(10))
print(mymath.is_prime(97))
print(mymath.primes_up_to(50))
```

## 12. File Handling

### 12.1 Opening Files (`open`)

| Mode | Meaning |
|---|---|
| `'r'` | Read (default) |
| `'w'` | Write (overwrites) |
| `'a'` | Append |
| `'x'` | Create (fails if exists) |
| `'b'` | Binary mode (combine: `'rb'`, `'wb'`) |
| `'+'` | Read + Write |


```python
# ── Creating / Writing ────────────────────────────────
filepath = "./sample.txt"

f = open(filepath, "w")
f.write("Line 1: Hello, Python!\n")
f.write("Line 2: File I/O is easy.\n")
f.write("Line 3: Let's explore more.\n")
f.close()

print("File written.")
```

```python
# ── Reading ───────────────────────────────────────────
# read() — entire file as one string
f = open(filepath, "r")
content = f.read()
f.close()
print("=== read() ===")
print(content)

# readline() — one line at a time
f = open(filepath, "r")
print("=== readline() ===")
line = f.readline()
while line:
    print(repr(line))
    line = f.readline()
f.close()

# readlines() — list of all lines
f = open(filepath, "r")
lines = f.readlines()
f.close()
print("=== readlines() ===")
print(lines)
```

```python
# ── with statement (recommended — auto closes file) ───
with open(filepath, "r") as f:
    content = f.read()
print("Content length:", len(content))

# ── readlines() + split ───────────────────────────────
with open(filepath, "r") as f:
    for line in f.readlines():
        parts = line.strip().split(": ")
        if len(parts) == 2:
            label, text = parts
            print(f"  {label!r:10} -> {text!r}")
```

```python
# ── Appending ─────────────────────────────────────────
with open(filepath, "a") as f:
    f.write("Line 4: Appended line.\n")
    f.write("Line 5: Another appended line.\n")

with open(filepath) as f:
    print(f.read())
```

```python
# ── Working with binary files ────────────────────────
bin_path = "./numbers.bin"

import struct
# Write 5 integers as binary
with open(bin_path, "wb") as f:
    for n in [1, 2, 3, 4, 5]:
        f.write(struct.pack("i", n))

# Read them back
with open(bin_path, "rb") as f:
    data = f.read()

nums = [struct.unpack("i", data[i:i+4])[0] for i in range(0, len(data), 4)]
print("Binary read back:", nums)

# ── Useful os.path helpers ────────────────────────────
import os
print("File size:", os.path.getsize(filepath), "bytes")
print("Exists:", os.path.exists(filepath))
print("Is file:", os.path.isfile(filepath))
```

## 13. Exception Handling

Exceptions are errors that occur at runtime. Python's `try/except` block allows you to handle them gracefully.


```python
# ── Basic try / except ───────────────────────────────
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Caught: {e}")

# ── Multiple except clauses ───────────────────────────
def safe_parse(value):
    try:
        return int(value)
    except ValueError:
        print(f"ValueError: '{value}' is not an integer")
    except TypeError:
        print(f"TypeError: cannot convert {type(value).__name__}")
    return None

print(safe_parse("42"))
print(safe_parse("abc"))
print(safe_parse(None))
```

```python
# ── else / finally ───────────────────────────────────
def read_file(path):
    try:
        f = open(path, "r")
    except FileNotFoundError:
        print(f"File not found: {path}")
    else:
        # Runs only if NO exception was raised
        content = f.read()
        f.close()
        print(f"Read {len(content)} chars")
        return content
    finally:
        # ALWAYS runs (cleanup, close connections, etc.)
        print("Finished attempt.")

read_file("./sample.txt")
print()
read_file("./nonexistent.txt")
```

```python
# ── Custom Exceptions ────────────────────────────────
class InsufficientFundsError(Exception):
    """Raised when withdrawal exceeds balance."""
    def __init__(self, amount, balance):
        self.amount  = amount
        self.balance = balance
        super().__init__(f"Cannot withdraw ${amount:.2f}: balance is ${balance:.2f}")


class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner   = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self.balance += amount
        print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(amount, self.balance)
        self.balance -= amount
        print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")


account = BankAccount("Alice", 100.0)
account.deposit(50)

try:
    account.withdraw(200)
except InsufficientFundsError as e:
    print(f"Error: {e}")

account.withdraw(75)
```

```python
# ── Common Built-in Exceptions ───────────────────────
exceptions_demo = [
    ("IndexError",   lambda: [][0]),
    ("KeyError",     lambda: {}["missing"]),
    ("TypeError",    lambda: "str" + 1),
    ("AttributeError", lambda: (42).upper()),
    ("NameError",    lambda: undefined_var),
    ("OverflowError",lambda: float("inf") + float("inf")),
]

for name, fn in exceptions_demo:
    try:
        fn()
    except Exception as e:
        print(f"{name:20}: {e}")
```
